# communication-node-placement-optimizer

Run this learning prototype in Google Colab. Execute the cells in order. Upload **communication-node-placement-optimizer.zip** in the next cell. No GitHub token or API key is needed.

Türkçe: Hücreleri sırayla çalıştır; bir sonraki hücrede yalnızca bu projenin ZIP'ini seç. Bu notebook aynı `main.py` dosyasını çalıştırır.


```python
from google.colab import files
from pathlib import Path
import io
import zipfile
import subprocess
import sys

uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError("Select exactly one project ZIP file.")
archive_name, archive_bytes = next(iter(uploaded.items()))
if not archive_name.lower().endswith(".zip"):
    raise ValueError("Upload the project ZIP, not an individual source file.")
base = Path("/content/engineering_projects")
base.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(archive_bytes)) as archive:
    names = archive.namelist()
    if "communication-node-placement-optimizer/main.py" not in names:
        raise ValueError("This is not the expected project ZIP.")
    for member in archive.infolist():
        if not member.filename.startswith("communication-node-placement-optimizer/"):
            continue
        target = (base / member.filename).resolve()
        if not target.is_relative_to(base.resolve()):
            raise ValueError("Unsafe archive path.")
        archive.extract(member, base)
project_dir = base / "communication-node-placement-optimizer"
print("Project extracted:", project_dir)
```

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-r", str(project_dir / "requirements.txt")], check=True)
```

```python
subprocess.run([sys.executable, "-m", "unittest", "-v"], cwd=project_dir, check=True)
```

## Run the model
Change the values below to explore the model. The default values reproduce the documented example.
Türkçe: Sayıları değiştirebilirsin; ilk çalıştırmada varsayılanları koru.


```python
grid_size = 30
nodes = 5
radius = 6.0
iterations = 1500
seed = 42
subprocess.run([sys.executable, "main.py", "--grid-size", str(grid_size),
    "--nodes", str(nodes), "--radius", str(radius), "--iterations", str(iterations),
    "--seed", str(seed), "--output-dir", "colab_results"], cwd=project_dir, check=True)
```
